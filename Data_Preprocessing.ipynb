```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import plotly.io as pio
```

```python
dataset=pd.read_csv("Lead_Scoring_Regression.csv")
```

```python
dataset
```

```python
print("Number of Rows:", dataset.shape[0])
print("Number of Columns:", dataset.shape[1])
```

```python
dataset.isnull().sum()
```

```python
dataset['Total_Visits'] = dataset['Total_Visits'].fillna('NA')
dataset['Time_Spent_Website'] = dataset['Time_Spent_Website'].fillna('NA')
dataset['Page_Views'] = dataset['Page_Views'].fillna('NA')
dataset['Email_Opens'] = dataset['Email_Opens'].fillna('NA')
dataset['Lead_Quality_Score'] = dataset['Lead_Quality_Score'].fillna('NA')
dataset['Annual_Income'] = dataset['Annual_Income'].fillna('NA')
```

```python
dataset.isnull().sum()
```

```python
dataset.to_csv("Lead_Scoring_Preprocessed.csv", index=False)
```

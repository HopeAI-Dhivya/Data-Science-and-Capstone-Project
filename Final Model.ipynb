```python
import pandas as pd
import pickle
```

```python
# Load model
model = pickle.load(open("Lead_Scoring_Linear_Regression_8Features.pkl", "rb"))

# Load scaler
scaler = pickle.load(open("Lead_Scoring_Scaler_8Features.pkl", "rb"))
```

```python
selected_features = [
    'Total_Visits',
    'Time_Spent_Website',
    'Email_Clicks',
    'Previous_Interactions',
    'Lead_Quality_Score',
    'Annual_Income',
    'Sales_Calls',
    'Days_Since_Last_Activity'
]
```

```python
# Create multiple new leads
new_leads = pd.DataFrame([
    {
        'Total_Visits': 15,
        'Time_Spent_Website': 650,
        'Email_Clicks': 5,
        'Previous_Interactions': 12,
        'Lead_Quality_Score': 75,
        'Annual_Income': 45000,
        'Sales_Calls': 10,
        'Days_Since_Last_Activity': 8
    },
    {
        'Total_Visits': 20,
        'Time_Spent_Website': 800,
        'Email_Clicks': 8,
        'Previous_Interactions': 15,
        'Lead_Quality_Score': 85,
        'Annual_Income': 60000,
        'Sales_Calls': 12,
        'Days_Since_Last_Activity': 5
    },
    {
        'Total_Visits': 10,
        'Time_Spent_Website': 400,
        'Email_Clicks': 3,
        'Previous_Interactions': 7,
        'Lead_Quality_Score': 60,
        'Annual_Income': 35000,
        'Sales_Calls': 4,
        'Days_Since_Last_Activity': 20
    },
    {
        'Total_Visits': 40,
        'Time_Spent_Website': 200,
        'Email_Clicks': 5,
        'Previous_Interactions': 8,
        'Lead_Quality_Score': 90,
        'Annual_Income': 40000,
        'Sales_Calls': 6,
        'Days_Since_Last_Activity': 15
    }
])
```

```python
# Selected exactly the 8 features used during training
new_leads = new_leads[selected_features]
```

```python
# Scaling
new_leads_scaled = scaler.transform(new_leads)
```

```python
# Prediction
prediction = model.predict(new_leads_scaled)
```

```python
# Display all predictions
for i, value in enumerate(prediction):
    print(f"Prediction [{i}] - Estimated Lead Value: {value:.3f}")
```

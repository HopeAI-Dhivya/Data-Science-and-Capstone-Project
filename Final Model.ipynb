```python
import pandas as pd
import pickle
```

```python
# Load model
model = pickle.load(open("Lead_Scoring_Linear_Regression_8Features.pkl", "rb"))

# Load scaler
scaler = pickle.load(open("Lead_Scoring_Scaler_8Features.pkl", "rb"))
```

```python
new_lead = pd.DataFrame([{
    'Total_Visits': 15,
    'Time_Spent_Website': 650,
    'Email_Clicks': 5,
    'Previous_Interactions': 12,
    'Lead_Quality_Score': 75,
    'Annual_Income': 45000,
    'Sales_Calls': 10,
    'Days_Since_Last_Activity': 8
}])
```

```python
selected_features = [
    'Total_Visits',
    'Time_Spent_Website',
    'Email_Clicks',
    'Previous_Interactions',
    'Lead_Quality_Score',
    'Annual_Income',
    'Sales_Calls',
    'Days_Since_Last_Activity'
]
```

```python
new_lead = new_lead[selected_features]
```

```python
# Scale new lead
new_lead_scaled = scaler.transform(new_lead)
```

```python
# Prediction
prediction = model.predict(new_lead_scaled)

print("Estimated Lead Value:", prediction[0])
```
